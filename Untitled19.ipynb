# Network Traffic Analysis with Scapy

In network security, analyzing packet captures (`.pcap` files) is a fundamental skill for diagnosing network behaviors, identifying anomalies, and tracing security incidents.

### Scapy Overview
[Scapy](https://scapy.net/) is a powerful Python-based interactive packet manipulation program and library. It can forge or decode packets of a wide number of protocols, send them on the wire, capture them, match requests and replies, and read/write PCAP files.

### Objectives
1. Install Scapy securely.
2. Generate a mock offline PCAP file featuring common protocols (TCP, UDP, DNS, HTTP).
3. Read, parse, and analyze the packet header attributes (Source IP, Destination IP, Protocol) from the file.

Now we will write a Python script that creates a mock PCAP file containing simulated network traffic, loads it using Scapy, and prints out the critical fields of each packet.

```python
from scapy.all import IP, TCP, UDP, DNS, DNSQR, wrpcap, rdpcap
import pandas as pd

# 1. Generate a dummy PCAP file with representative packets
def generate_dummy_pcap(filename):
    packets = [
        # HTTP-like TCP handshake/data
        IP(src="192.168.1.50", dst="93.184.216.34")/TCP(sport=49152, dport=80, flags="S"),
        IP(src="93.184.216.34", dst="192.168.1.50")/TCP(sport=80, dport=49152, flags="SA"),
        IP(src="192.168.1.50", dst="93.184.216.34")/TCP(sport=49152, dport=80, flags="A"),

        # DNS Query over UDP
        IP(src="192.168.1.50", dst="8.8.8.8")/UDP(sport=53530, dport=53)/DNS(rd=1, qd=DNSQR(qname="example.com")),
        IP(src="8.8.8.8", dst="192.168.1.50")/UDP(sport=53, dport=53530)/DNS(qr=1, an=None),

        # Local loopback test
        IP(src="127.0.0.1", dst="127.0.0.1")/TCP(sport=8080, dport=50000, flags="PA")
    ]
    wrpcap(filename, packets)
    print(f"[+] Successfully created dummy PCAP file: '{filename}' containing {len(packets)} packets.\n")

# 2. Parse and analyze the PCAP file
def analyze_pcap(filename):
    print(f"[*] Loading and parsing {filename}...")
    packets = rdpcap(filename)

    packet_data = []

    # Protocol mapping dictionary
    proto_map = {6: "TCP", 17: "UDP", 1: "ICMP"}

    for idx, packet in enumerate(packets):
        if IP in packet:
            src_ip = packet[IP].src
            dst_ip = packet[IP].dst
            proto_num = packet[IP].proto
            protocol = proto_map.get(proto_num, f"Other ({proto_num})")

            # Identify application layer if available
            info = ""
            if packet.haslayer(TCP):
                info = f"TCP Port: {packet[TCP].sport} -> {packet[TCP].dport} (Flags: {packet[TCP].flags})"
            elif packet.haslayer(UDP):
                info = f"UDP Port: {packet[UDP].sport} -> {packet[UDP].dport}"
                if packet.haslayer(DNS):
                    info += " (DNS Query)"

            packet_data.append({
                "Packet #": idx + 1,
                "Source IP": src_ip,
                "Destination IP": dst_ip,
                "Protocol": protocol,
                "Details": info
            })

    # Present findings clearly in a pandas DataFrame
    df = pd.DataFrame(packet_data)
    display(df)

# Run the creation and the analysis
pcap_file = "simulated_traffic.pcap"
generate_dummy_pcap(pcap_file)
analyze_pcap(pcap_file)
```
