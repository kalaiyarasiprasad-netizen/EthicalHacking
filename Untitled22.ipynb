### Packet Analyser Tool using Scapy

This tool parses network packet capture files (`.pcap`) and extracts fundamental diagnostic metadata from the IP layer, including packet numbers, source/destination IPs, protocol type, and exact packet sizes (length).

```python
from scapy.all import rdpcap, IP
import pandas as pd
import os

def analyze_packets(pcap_path):
    # Ensure the file exists (fallback to generating one if needed)
    if not os.path.exists(pcap_path):
        print(f"[-] '{pcap_path}' not found. Generating a new sample capture...")
        from scapy.all import IP, TCP, UDP, wrpcap
        sample_packets = [
            IP(src="192.168.1.10", dst="8.8.8.8")/UDP(sport=53, dport=53),
            IP(src="8.8.8.8", dst="192.168.1.10")/UDP(sport=53, dport=53),
            IP(src="192.168.1.10", dst="142.250.190.46")/TCP(sport=443, dport=443, flags="S"),
            IP(src="142.250.190.46", dst="192.168.1.10")/TCP(sport=443, dport=443, flags="SA")
        ]
        wrpcap(pcap_path, sample_packets)

    print(f"[*] Loading PCAP file: '{pcap_path}'")
    packets = rdpcap(pcap_path)

    analysis_results = []
    protocol_mapping = {1: "ICMP", 6: "TCP", 17: "UDP"}

    print(f"[*] Parsing {len(packets)} packets...")
    for index, packet in enumerate(packets, start=1):
           if IP in packet:
            src_ip = packet[IP].src
            dst_ip = packet[IP].dst
            proto_num = packet[IP].proto
            protocol_name = protocol_mapping.get(proto_num, f"RAW ({proto_num})")
            packet_length = len(packet)  # Retrieves size of the packet in bytes

            analysis_results.append({
                "Packet Number": index,
                "Source IP": src_ip,
                "Destination IP": dst_ip,
                "Protocol": protocol_name,
                "Length (Bytes)": packet_length
            })

    # Display findings in a structured table
    df = pd.DataFrame(analysis_results)
    display(df)

# Run the analyzer tool
analyze_packets("simulated_traffic.pcap")
```
