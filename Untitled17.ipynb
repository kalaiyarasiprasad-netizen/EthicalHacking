# Windows & Linux OS Network Diagnostics Commands

Network administrators and ethical hackers use standard OS commands to diagnose network connectivity, analyze routing paths, check active connections, and examine adapter configurations.

* **Ping**: Tests host reachability and round-trip time.
* **Traceroute / Tracert**: Maps the path network packets take to a destination.
* **Ipconfig / Ifconfig**: Displays network adapter configurations.
* **Netstat**: Shows active network connections, routing tables, and interface statistics.

```python
import subprocess
import platform

def run_command(command_list):
    try:
        result = subprocess.run(command_list, stdout=subprocess.PIPE, stderr=subprocess.PIPE, text=True, timeout=10)
        if result.returncode == 0:
            return result.stdout
        else:
            return f"Error executing command:\n{result.stderr}"
    except Exception as e:
        return f"Execution failed: {str(e)}"

os_type = platform.system()
print(f"[*] Detected Operating System: {os_type}\n")

# 1. Ping
print("=== 1. Ping Command (Target: localhost) ===")
ping_cmd = ["ping", "-n", "3", "127.0.0.1"] if os_type == "Windows" else ["ping", "-c", "3", "127.0.0.1"]
print(run_command(ping_cmd))

# 2. Ipconfig / Ifconfig
print("=== 2. IP Configuration Command ===")
ip_cmd = ["ipconfig"] if os_type == "Windows" else ["ifconfig"]
print(run_command(ip_cmd))

# 3. Traceroute
print("=== 3. Traceroute Command (Target: localhost) ===")
trace_cmd = ["tracert", "127.0.0.1"] if os_type == "Windows" else ["traceroute", "127.0.0.1"]
print(run_command(trace_cmd))

# 4. Netstat
print("=== 4. Netstat Command ===")
net_cmd = ["netstat", "-an"] if os_type == "Windows" else ["netstat", "-ant"]
# Fetching first 15 lines of netstat output for display brevity
netstat_output = run_command(net_cmd)
print("\n".join(netstat_output.splitlines()[:15]))
```
