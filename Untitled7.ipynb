### 1. `uptime` Command in Ethical Hacking
The `uptime` command tells you how long the target server has been running, the number of active users, and the system load averages. In penetration testing, high uptime indicates a target machine that has likely not been patched or rebooted recently, increasing the probability of kernel-level vulnerabilities being active.

### 3. `mv` Command in Ethical Hacking
`mv` (Move) is used to rename or shift files around the filesystem. In exploitation, `mv` can be used to hide tools inside deep directories, rename standard shell scripts to innocent configuration filenames to evade suspicion, or stage files for extraction.

```python
# Create a simulated tool script file
with open("backdoor_shell.sh", "w") as f:
    f.write("#/bin/bash\n# Simulation of payload\necho 'Payload Active'\n")

# Evade quick visual checks by moving/renaming it to look like a configuration backup
!mv backdoor_shell.sh backup_cron_service.conf

# Verify that the file was renamed and moved
!ls -la backup_cron_service.conf
```
