## 1. Nmap Ping Scan (`-sn`)

Also known as a 'no-port scan', this instructs Nmap to determine active hosts on the subnet without scanning individual ports.

```python
# Run a ping scan on localhost to check if the host is active
!nmap -sn 127.0.0.1
```

## 2. ICMP Echo Request (`-PE`)

Sends an ICMP Echo Request to determine if a host is alive, resembling a traditional ping but integrated into the Nmap framework.

```python
# Perform an ICMP Echo Request host discovery on localhost
!nmap -PE -sn 127.0.0.1
```
