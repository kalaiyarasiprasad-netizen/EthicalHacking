### Nmap Scans for Ethical Hacking
Nmap (Network Mapper) is an open-source tool used for network discovery and security auditing. In Kali Linux, it is one of the most fundamental tools for reconnaissance.

#### i) Detect the Version of Services Running (`-sV`)
This command probes open ports to determine service and version info. Identifying the exact version is critical to finding known vulnerabilities associated with that service.

```python
# Execute a service version detection scan on localhost
!nmap -sV localhost
```

#### ii) Aggressive Scan (`-A`)
An aggressive scan enables OS detection (`-O`), service version detection (`-sV`), script scanning (`-sC`), and traceroute. It is very loud on the network but gathers comprehensive details quickly.

```python
# Execute an aggressive scan on localhost
!nmap -A localhost
```

#### iii) Detect Operating System of the Target (`-O`)
This sends out TCP and UDP packets to the target and analyzes the responses against a database of known OS fingerprints to determine the operating system.

```python
# Execute OS detection on localhost (requires root privileges on native Linux systems)
!sudo nmap -O localhost
```
