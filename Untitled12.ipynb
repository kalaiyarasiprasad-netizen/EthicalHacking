### Nmap Performance and Speed Templates

Nmap timing templates (`-T0` through `-T5`) control the speed and parallelization of scans. When scanning speeds are increased, the scan becomes faster but much louder and more prone to missing open ports or being blocked by defensive security controls.

* **Aggressive (`-T4`)**: Speeds up scans, assuming you are on a reasonably fast and reliable network. Commonly used in CTFs and authorized pentests.
* **Insane (`-T5`)**: Extremely fast scan. It is highly likely to drop packets and trigger IDS unless you have massive bandwidth and a highly robust target network.

#### (i) Aggressive Speed Scan (`-T4`)
An aggressive timing template optimizing delays and timeouts for rapid scanning.

```python
# Run an aggressive performance scan on localhost
!nmap -T4 localhost
```

#### (ii) Insane Speed Scan (`-T5`)
An extremely rapid template designed for speed, capping round-trip times to minimum values.

```python
# Run an insane performance scan on localhost
!nmap -T5 localhost
```
