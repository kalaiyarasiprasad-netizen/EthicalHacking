### Windows Batch File Execution Demonstration

This notebook demonstrates how to programmatically create and execute a Windows batch file (`.bat`) using Python.

Since Google Colab runs on a **Linux** runtime, executing a Windows batch script directly via a shell sub-process is not natively supported. To solve this, our Python program will:
1. Create the harmless Windows batch file `demo.bat` with standard commands (`@echo off`, `echo`, `date`, `time`).
2. Detect the host Operating System.
3. Run the batch file natively if on Windows, or parse and run its logical equivalent using Linux-compatible commands if running in Colab.

```python
import os
import platform
import subprocess

# Define the filename and content of our harmless batch file
bat_filename = "demo.bat"
bat_content = """@echo off
echo Hello World!
echo Current Date:
date /t
echo Current Time:
time /t
"""

# Write the content to a .bat file
with open(bat_filename, "w") as bat_file:
    bat_file.write(bat_content)

print(f"[+] Created batch file '{bat_filename}' with the following content:")
print("-" * 40)
print(bat_content.strip())
print("-" * 40)

# Detect OS and execute accordingly
current_os = platform.system()
print(f"\n[*] Detected Operating System: {current_os}")

if current_os == "Windows":
    print("[*] Running batch file natively on Windows...")
    try:
        # Run the batch file and capture the standard output
        result = subprocess.run([bat_filename], capture_output=True, text=True, shell=True, check=True)
        print("\n[+] Execution Output:")
        print("-" * 40)
        print(result.stdout)
        print("-" * 40)
    except Exception as e:
        print(f"[!] Failed to execute native batch file: {e}")
else:
    print("[*] Running on Linux/Colab. Simulating/translating batch execution...")
    print("\n[+] Simulated Execution Output:")
    print("-" * 40)

    # Simulate '@echo off'
    # Simulate 'echo Hello World!'
    print("Hello World!")

    # Simulate 'echo Current Date:' & 'date /t'
    print("Current Date:")
    # Fetch and format date from Linux environment to match Windows style (e.g., Mon 10/25/2023)
    linux_date = subprocess.run(["date", "+%a %m/%d/%Y"], capture_output=True, text=True).stdout.strip()
    print(linux_date)

    # Simulate 'echo Current Time:' & 'time /t'
    print("Current Time:")
    # Fetch and format time from Linux environment to match Windows style (e.g., 02:30 PM)
    linux_time = subprocess.run(["date", "+%I:%M %p"], capture_output=True, text=True).stdout.strip()
    print(linux_time)
    print("-" * 40)
```
