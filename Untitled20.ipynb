```python
class SimulatedDisk:
    def __init__(self):
        # Initialize a simulated disk containing 10 sectors of 512 bytes each
        self.sectors = {i: bytearray(512) for i in range(10)}
        self.setup_clean_system()

    def setup_clean_system(self):
        # Setup Sector 0 (MBR) with benign, original bootstrap instructions
        orig_bootstrap = b"[ORIGINAL BOOT LOADER: Load OS Kernel into Memory]"
        self.sectors[0][:len(orig_bootstrap)] = orig_bootstrap
        # Add partition table placeholders and boot signature
        self.sectors[0][510:512] = b"\x55\xAA"

        print("[*] Clean system initialized.")
        print(f"    Sector 0 (MBR) payload: {self.sectors[0][:50].decode('utf-8', errors='ignore')}...")

    def read_sector(self, num):
        return self.sectors[num]

    def write_sector(self, num, data):
        if len(data) != 512:
            raise ValueError("Sectors must be exactly 512 bytes.")
        self.sectors[num] = bytearray(data)


def simulate_boot_sector_infection(disk):
    print("\n--- Step 1: Simulating Virus Attempting Infection ---")

    # Read the original MBR (Sector 0)
    original_mbr = disk.read_sector(0)

    # Prepare the mock virus payload (strictly educational message)
    virus_payload = bytearray(512)
    malicious_routine = b"[SIMULATED VIRUS: Display warning; then pass control]"
    virus_payload[:len(malicious_routine)] = malicious_routine
    # Preserve the partition table layout & signature from original MBR
    virus_payload[446:] = original_mbr[446:]

    # The historic 'Stealth' method: Relocate the original MBR to an 'unused' sector (e.g., Sector 9)
    print("[*] Copying original boot loader to safe sector (Sector 9)...")
    disk.write_sector(9, original_mbr)

    # Overwrite the main MBR (Sector 0) with the virus code
    print("[*] Overwriting main MBR (Sector 0) with simulated virus vector...")
    disk.write_sector(0, virus_payload)
    print("[+] Infection routine completed (in simulation).\n")


def simulate_system_boot(disk):
    print("--- Step 2: Simulating System Boot Cycle ---")
    print("[*] BIOS reads Sector 0 (MBR) and executes instructions...")

    mbr = disk.read_sector(0)
    execution_pointer = mbr[:100].decode('utf-8', errors='ignore').strip('\x00')
    print(f"\n[EXECUTION] Primary MBR executing code: \n>>> \"{execution_pointer}\"")

    if "SIMULATED VIRUS" in execution_pointer:
        print("\n[!] ALERT: Simulated payload active in memory.")
        print("[*] Virus routine completes. Redirecting execution back to relocated original loader...")

        # Read and execute relocated original MBR from Sector 9
        relocated_mbr = disk.read_sector(9)
        original_pointer = relocated_mbr[:100].decode('utf-8', errors='ignore').strip('\x00')
        print(f"\n[EXECUTION] Handing off control back to: \n>>> \"{original_pointer}\"")
        print("\n[+] Operating System loaded successfully despite simulated infection detour.")
    else:
        print("\n[+] System booted safely and directly to Operating System.")

# --- Execution of the Demonstration ---
virtual_drive = SimulatedDisk()

# Simulate normal boot of a healthy drive
simulate_system_boot(virtual_drive)

# Perform simulated legacy infection redirection
simulate_boot_sector_infection(virtual_drive)

# Simulate subsequent boot of infected drive to inspect redirection behavior
simulate_system_boot(virtual_drive)
```
