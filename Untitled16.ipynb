# WHOIS Domain Reconnaissance

WHOIS is a query and response protocol widely used for querying databases that store the registered users or assignees of an Internet resource, such as a domain name or an IP address block.

```python
!pip install python-whois
```

```python
import whois

target_domain = "example.com"
print(f"[*] Performing WHOIS lookup for: {target_domain}\n")

try:
    w = whois.whois(target_domain)

    # Display specified parameters
    print(f"Domain Name: {w.domain_name}")
    print(f"Registrar: {w.registrar}")
    print(f"Creation Date: {w.creation_date}")
    print(f"Expiration Date: {w.expiration_date}")
except Exception as e:
    print(f"[!] Error retrieving WHOIS information: {e}")
```
