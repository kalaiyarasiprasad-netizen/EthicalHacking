### 1. `ls` Command (Listing Files with Security Context)
In an audit or investigation, standard `ls` is expanded with flags like `-l` (long formatting) to verify file permissions (e.g., checking if critical files are globally writeable) and `-a` (showing hidden files like `.bash_history` or `.env` files).

```python
# List the current directory contents showing all permissions, owners, and hidden items
!ls -la
```

### 2. `cat` Command (Concatenate/Read Files)
Ethical hackers use `cat` to quickly view the contents of system files (like `/etc/passwd`), custom exploit configurations, or logs. Combining `cat` with pipes allows for quick filtering of sensitive patterns like IP addresses or database credentials.

```python
# Read the contents of our previously generated security file
!cat services.conf | head -n 10
```

### 3. `mkdir` Command (Make Directory)
During post-exploitation or a system audit, analysts use `mkdir` to set up working environments, download directories for tools, or structure collected exfiltration data securely.

```python
# Create a dedicated workspace folder for security auditing
!mkdir -p workspace/exfil_data

# Confirm directory creation
!ls -ld workspace/exfil_data
```
