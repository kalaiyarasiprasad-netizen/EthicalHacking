# Port Scanning Tools (Ethical Hacking)

Port scanning is a reconnaissance technique used to identify open ports and services running on a host. In addition to command-line utilities like Nmap, ethical hackers often write custom scripts in Python to perform stealthy, lightweight, or custom scans.

## 1. Custom Single-Threaded TCP Port Scanner
Below is a simple Python scanner using the standard socket library to scan common ports on localhost.

```python
import socket
import sys
from datetime import datetime

target = "127.0.0.1"
ports_to_scan = [21, 22, 23, 25, 53, 80, 110, 443, 3306, 8080]

print(f"[*] Starting scan on host: {target}")
print(f"[*] Time started: {str(datetime.now())}\n")

for port in ports_to_scan:
    s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    s.settimeout(0.5)  # Quick timeout to prevent long delays
    result = s.connect_ex((target, port))
    if result == 0:
        print(f"[+] Port {port}: OPEN")
    else:
        print(f"[-] Port {port}: CLOSED")
    s.close()
```
