### Resource Monitoring and Command History (Ethical Hacking Utilities)

We will now run commands to inspect memory utilization, parse and sort targets or log data, and review command histories.

#### i) Free Command
Displays the total amount of free and used physical and swap memory in the system.

```python
# Show system memory utilization in a human-readable format
!free -h
```

#### ii) Sort Command
Used to sort lines of text files. This is extremely helpful during recon to sort unique IP addresses, subdomains, or scanned ports.

```python
# Create a quick list of unsorted subdomains and sort them alphabetically
with open('subdomains.txt', 'w') as f:
    f.write('target.com\nadmin.target.com\napi.target.com\nblog.target.com\n')

!sort subdomains.txt
```

#### iii) History Command
Displays the command history list with line numbers. (Note: Inside a non-interactive shell like Jupyter, we simulate history by reading the shell history file directly).

```python
# Read the shell command history from the current session history file if available
!history | tail -n 10 || echo 'No interactive bash history available in this subshell.'
```
