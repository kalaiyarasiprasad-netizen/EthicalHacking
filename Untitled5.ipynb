### 1. `cd` Command
`cd` (Change Directory) allows an auditor or attacker to traverse the directory tree. In post-exploitation, moving to system directories (like `/etc` or `/var/log`) is key to discovering sensitive files or system configurations.

```python
# Simulate directory change and path verification
!mkdir -p target_logs
!cd target_logs && pwd
```

### 2. `cp` Command
In security operations, `cp` (copy) is used to duplicate sensitive assets before editing them (to keep a backup) or to stage target files (like database files or configuration profiles) into a readable public folder or a web root.

```python
# Copy a sensitive local file into our staging log directory for collection
!cp /content/services.conf target_logs/

# Confirm that the file has been successfully copied
!ls target_logs
```

### 3. `ls` Command
`ls` (List) displays directory contents. When executing an audit, flags like `-l` (long listing), `-a` (show hidden files like `.bash_history` or `.git`), and `-t` (sort by modification time to see recently edited scripts) provide valuable intelligence.

```python
# List target_logs contents with detailed permissions, ownership, and hidden files
!ls -la target_logs
```
