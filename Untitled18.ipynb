### Educational Web Vulnerability Analysis (Nikto Simulation)

In professional environments, tools like **Nikto** are used to perform rapid diagnostic scans against web servers. They check for:
1. **Default or Outdated Files**: Known default configurations, sample files, or administrative panels left open.
2. **Interesting/Exposed Headers**: Server banners exposing specific software versions or missing essential security headers (like `X-Frame-Options` or `Content-Security-Policy`).
3. **Common CGI Vulnerabilities**: Outdated scripts or paths known to contain exploitable flaws (e.g., shellshock or unauthenticated access).

Below is a safe, self-contained Python script simulating how an automated scanner checks paths, inspects security headers, and reports potential misconfigurations.

```python
import urllib3
import json

# Configure a mock target representation of a misconfigured local server for demonstration
mock_headers = {
    "Server": "Apache/2.4.41 (Ubuntu) PHP/7.3.11",
    "X-Powered-By": "PHP/7.3.11",
}

# List of common paths and files a CGI/web scanner looks for
scan_database = [
    {
        "path": "/cgi-bin/test-cgi",
        "description": "Legacy test CGI script",
        "expected_impact": "Information Disclosure / Potential Remote Code Execution"
    },
    {
        "path": "/admin/",
        "description": "Unprotected Administrative Interface",
        "expected_impact": "Unauthorized administrative access"
    },
    {
        "path": "/config.json.bak",
        "description": "Exposed backup configuration files",
        "expected_impact": "Credential leak & sensitive database information"
    },
    {
        "path": "/phpinfo.php",
        "description": "Exposed PHP configuration info",
        "expected_impact": "Information leakage of environment variables"
    }
]

def analyze_security_headers(headers):
    """Analyze server HTTP headers for missing security configurations."""
    findings = []
    security_headers = [
        "X-Frame-Options",
        "Content-Security-Policy",
        "X-Content-Type-Options",
        "Strict-Transport-Security"
    ]

    # Check banner disclosure
    server_banner = headers.get("Server")
    if server_banner:
        findings.append({
            "Type": "Header Disclosure",
            "Details": f"Detailed Server header leaked: '{server_banner}' (Reveals operating system & exact version)",
            "Severity": "Low"
        })

    for sec_header in security_headers:
        if sec_header not in headers:
            findings.append({
                "Type": "Missing Security Header",
                "Details": f"Missing crucial security header: '{sec_header}'",
                "Severity": "Medium"
            })

    return findings

def run_web_assessment_simulation():
    print("=" * 75)
    print("        EDUCATIONAL WEB VULNERABILITY & CGI SCANNER SIMULATION         ")
    print("=" * 75)

    # 1. Analyze Header Configurations
    print("\n[*] Phase 1: Analyzing HTTP Response Headers...")
    header_findings = analyze_security_headers(mock_headers)
    for finding in header_findings:
        print(f"  [!] [{finding['Severity']}] {finding['Type']}: {finding['Details']}")

    # 2. Simulate Path Scans (identifying common exposed locations)
    print("\n[*] Phase 2: Scanning common paths / CGI directories...")
    for item in scan_database:
        # Simulating that we find these configuration issues locally
        print(f"  [+] Testing path: {item['path']} -> FOUND (Status Code: 200 OK)")
        print(f"      └─ Description: {item['description']}")
        print(f"      └─ Risk: {item['expected_impact']}")
        print("-" * 60)

    print("\n[+] Assessment Complete. Recommended Actions:")
    print("    1. Remove any .bak, .old, or temporary setup files from the production web root.")
    print("    2. Strip down the 'Server' response header configuration to hide exact software versions.")
    print("    3. Restrict /cgi-bin/ directories to specific internal IPs and disable execute permissions on unused test files.")
    print("=" * 75)

run_web_assessment_simulation()
```
