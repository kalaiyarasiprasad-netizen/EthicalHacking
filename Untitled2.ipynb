### 3. `vi` Command
`vi` is a terminal-based text editor found on almost all Unix-like systems. Ethical hackers use it to edit local payloads, alter target server configurations, or modify scripts. Because `vi` is interactive, we will demonstrate how to write a file scriptably using EX command mode commands (non-interactive editing).

```python
# Create a quick reverse shell python script placeholder
with open("payload.py", "w") as f:
    f.write("import os\nimport sys\nprint('Staging payload...')\n")

# Let's use vi non-interactively to append a line of code at the end of the script using Ex commands (-c)
!vi -c "$s/$/\nprint('Payload Executed Successfully!')/" -c "wq" payload.py

# Print the result to verify vi's modification
!cat payload.py
```
