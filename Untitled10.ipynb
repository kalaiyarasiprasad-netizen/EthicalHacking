### Nmap Timing Templates for IDS Evasion

Nmap offers timing templates (`-T0` through `-T5`) to speed up or slow down scans. Adjusting these templates is crucial for evasion and minimizing target impact during ethical hacking engagements:
- **Polite (`-T1`)**: Slows down the scan to use less bandwidth and evade Intrusion Detection Systems (IDS).
- **Normal (`-T3`)**: The default timing profile, which behaves normally without parallelization adjustments.

#### i) Polite IDS Evasion Scan (`-T1`)
This timing option serializes scans and inserts a minimum 15-second delay between probes to evade detection rules.

```python
# Execute a polite timing template scan on localhost to demonstrate IDS evasion behavior
!nmap -T1 localhost
```

#### ii) Normal IDS Evasion / Default Scan (`-T3`)
This timing option runs probes dynamically with parallel tasks and is the default behavior if no template is specified.

```python
# Execute a normal timing template scan on localhost
!nmap -T3 localhost
```
