### 1. `touch` Command
In ethical hacking, `touch` is commonly used to pre-create log files, empty payload scripts, or decoy files on a target system. It can also modify timestamps to match existing files (timestomping) to blend in with legitimate OS files.

```python
# Create a decoy system configuration file using touch
!touch decoy_config.conf

# List the newly created file to verify
!ls -la decoy_config.conf
```

### 2. `pwd` Command
`pwd` (Print Working Directory) tells you exactly where you are in the filesystem. When a hacker gains initial shell access, executing `pwd` is one of the first reconnaissance actions taken to determine their entry point and map out relative directories.

```python
# Print current working directory to establish orientation
!pwd
```

### 3. `wget` Command
`wget` is a network utility used to retrieve files from web servers. In penetration testing, it is heavily used during the post-exploitation stage to download exploit scripts, tools (like LinPeas), or reverse shells directly onto a compromised machine from the attacker's web server.

```python
# Simulate downloading a safe sample file (the EICAR anti-malware test file) to demonstrate download behavior
!wget -q https://secure.eicar.org/eicar.com.txt -O safe_test_payload.txt

# Display the downloaded content
!cat safe_test_payload.txt
```
