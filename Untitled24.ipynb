### Nmap Timing and Performance (T0 to T5)

Nmap offers six timing templates (numbered 0 to 5) to control how aggressively and quickly a scan is executed. Adjusting these parameters helps optimize performance or evade detection (such as Intrusion Detection Systems):

| Template | Name | Inter-probe Delay | Main Use Case / Description |
| :--- | :--- | :--- | :--- |
| **T0** | Paranoid | 5 minutes | Serialized scanning designed to completely evade detection; extremely slow. |
| **T1** | Sneaky | 15 seconds | Avoids IDS alerts by spacing out packet transmissions. |
| **T2** | Polite | 0.4 seconds | Reduces bandwidth usage and load on the target server. |
| **T3** | Normal | 0 seconds (dynamic) | The default mode; behaves dynamically based on host responsiveness. |
| **T4** | Aggressive | 0 seconds (dynamic) | Sets strict, fast timeouts (max TCP round-trip timeout < 1.25s). Highly recommended on fast, reliable networks. |
| **T5** | Insane | 0 seconds (dynamic) | Extremely fast; drops accuracy and risks missing ports in favor of raw speed. |

```python
import socket
import time
import sys

def simulate_timing_scan(template_name, delay_seconds, ports=[22, 80, 443]):
    """Simulates a TCP Connect scan with specific inter-probe delays to replicate Nmap templates."""
    print(f"\n[*] Initializing simulated Nmap {template_name} scan against localhost (127.0.0.1)...")
    print(f"[*] Configured Inter-Probe Delay: {delay_seconds} seconds")
    print("-" * 60)

    start_time = time.time()
    discovered_ports = []

    for port in ports:
        # Replicating the inter-probe delay before scanning the next port
        if delay_seconds > 0:
            time.sleep(delay_seconds)

        s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
        s.settimeout(0.5)  # Quick timeout to represent optimized performance

        result = s.connect_ex(("127.0.0.1", port))
        if result == 0:
            discovered_ports.append(port)
            print(f"  [+] Port {port} is OPEN")
        else:
            print(f"  [-] Port {port} is CLOSED")

        s.close()

    end_time = time.time()
    elapsed_time = end_time - start_time
    print("-" * 60)
    print(f"[+] {template_name} scan finished.")
    print(f"[+] Total Execution Time: {elapsed_time:.4f} seconds")
    return elapsed_time

# Let's run a comparison between a Polite (T2-like delay) and an Aggressive (T4-like delay) approach.
# We will use short representative delays to keep execution fast in the notebook environment.
t2_time = simulate_timing_scan("T2 (Polite Simulation)", delay_seconds=0.4)
t4_time = simulate_timing_scan("T4 (Aggressive Simulation)", delay_seconds=0.0)
```
