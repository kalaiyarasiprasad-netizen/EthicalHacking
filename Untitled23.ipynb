### Educational Port Scanner Using Sockets

Port scanning is a fundamental networking technique to identify which services are actively listening on a host.

#### Socket-Based Scanning Method (`TCP Connect` Scan)
In Python, we can perform a standard TCP three-way handshake attempt using the `socket` module:
- **`socket.connect()`**: Throws an exception if the connection fails.
- **`socket.connect_ex()`**: Returns an error code directly instead of raising an exception. A return value of `0` indicates the socket successfully connected, meaning the port is **Open**. Any other value (such as `111` on Linux or `10061` on Windows) indicates the port is **Closed** or unreachable.

```python
import socket
import sys
from datetime import datetime

def run_simple_port_scanner(target_host="127.0.0.1"):
    # Define a list of well-known and common ports to check
    ports_to_check = [22, 80, 443, 8080, 9000, 2222, 3453, 6000]

    # Resolve target hostname to IP address safely
    try:
        target_ip = socket.gethostbyname(target_host)
    except socket.gaierror:
        print(f"[-] Hostname '{target_host}' could not be resolved.")
        return

    print("=" * 70)
    print(f"[*] Scanning Target : {target_host} ({target_ip})")
    print(f"[*] Time Started     : {str(datetime.now())}")
    print("=" * 70)

    for port in ports_to_check:
        # Set up a TCP socket block
        # AF_INET = IPv4, SOCK_STREAM = TCP
        s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
        s.settimeout(1.0) # Set timeout to 1 second to avoid hanging indefinitely

        # connect_ex returns 0 if connection was successful
        result = s.connect_ex((target_ip, port))

        if result == 0:
            print(f"[+] Port {port:5} : OPEN")
            print(f"    └─ Explanation: The port accepted the TCP handshake. An active service is listening here.")
        else:
            print(f"[-] Port {port:5} : CLOSED")
            print(f"    └─ Explanation: Connection refused/timed out (Error code {result}). No service is listening here.")

        s.close() # Always close sockets to release system resources

    print("=" * 70)
    print("[+] Scan completed successfully.")

# Run scanner on localhost
run_simple_port_scanner()
```
