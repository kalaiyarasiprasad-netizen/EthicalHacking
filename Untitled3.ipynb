### 1. `tree` Command in Ethical Hacking
During the enumeration and information-gathering phase, mapping the directory structure of an target system is vital. The `tree` command visually highlights directories and configuration file paths, helping an auditor quickly locate potential areas of vulnerability (such as configuration files, scripts, or sensitive storage).

```python
# Install tree utility
!apt-get install -y tree > /dev/null

# Create a mock target infrastructure directory for illustration
!mkdir -p victim_server/etc/ssl victim_server/var/www/html victim_server/home/database
!touch victim_server/etc/ssl/cert.pem victim_server/var/www/html/index.php victim_server/home/database/backup.sql

# Run the tree command to enumerate
!tree victim_server
```

### 2. `head` Command in Ethical Hacking
The `head` command displays the first few lines of a file (default is 10). Security engineers and penetration testers use `head` to quickly inspect file headers, analyze script entry points, or verify the structure of leaked credential lists without opening massive files in memory.

```python
# Generate a large file containing hypothetical logs or credential dumps
with open("compromised_hashes.txt", "w") as f:
    for i in range(1, 100):
        f.write(f"user_{i}:hash_value_{i * 12345}\n")

# Inspect only the top 5 records using head
!head -n 5 compromised_hashes.txt
```

### 3. `tail` Command in Ethical Hacking
The `tail` command outputs the last lines of a file (default is 10). In ethical hacking, blue teaming, or system administration, `tail -f` (follow mode) is crucial for real-time monitoring of authentication logs (`/var/log/auth.log`) to detect intrusion attempts or failed login activities as they happen.

```python
# Inspect the end of our hypothetical hash file using tail
!tail -n 5 compromised_hashes.txt
```
